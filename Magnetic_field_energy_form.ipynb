```python
import numpy as np
import matplotlib.pyplot as plt

# Inductances (henries)
L1 = 2.0
L2 = 3.0
M = 1.0

# Selected currents (amperes)
I1 = 2.0
I2 = 1.0

# Inductance matrix
A = np.array([
    [L1, M],
    [M, L2]
])

# Create a grid of possible currents
i1_values = np.linspace(-4, 4, 100)
i2_values = np.linspace(-4, 4, 100)

I1_grid, I2_grid = np.meshgrid(i1_values, i2_values)

# Calculate energy:
# W = 1/2 * (L1*I1^2 + 2*M*I1*I2 + L2*I2^2)
W_grid = 0.5 * (
    L1 * I1_grid**2
    + 2 * M * I1_grid * I2_grid
    + L2 * I2_grid**2
)

# Create the contour plot
plt.figure(figsize=(7, 5))

contour = plt.contourf(
    I1_grid,
    I2_grid,
    W_grid,
    levels=30
)

plt.colorbar(contour, label="Energy (J)")

# Mark the selected current point
plt.scatter(
    I1,
    I2,
    s=80,
    marker="x",
    label=f"Selected point ({I1}, {I2})"
)

plt.xlabel("I₁ (A)")
plt.ylabel("I₂ (A)")
plt.title("Magnetic Energy: W = ½ IᵀAI")
plt.legend()
plt.grid(True)

plt.show()
```
